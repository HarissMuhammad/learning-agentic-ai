```python
from agents import Agent, Runner, ItemHelpers, trace
import asyncio
```

```python
spanish_agent = Agent(
    name="Spanish Agent", instructions="You translate the given text to spanish"
)
translation_picker = Agent(
    name="Picker Agent",
    instructions="You pick the best Spanish translation from the given options.",
)


msg = input("Hi! Enter a message, and we'll translate it to Spanish.\n\n")

with trace("Parallel Translation"):
    res1, res2, res3 = await asyncio.gather(
        Runner.run(spanish_agent, msg),
        Runner.run(spanish_agent, msg),
        Runner.run(spanish_agent, msg),
    )

outputs = [
    ItemHelpers.text_message_outputs(res1.new_items),
    ItemHelpers.text_message_outputs(res2.new_items),
    ItemHelpers.text_message_outputs(res3.new_items),
]

translations = "\n\n".join(outputs)
print(f"\n\n Translations : {translations}\n\n")

best_translation = await Runner.run(translation_picker, f"input : {msg} {translations}")

print(f"best Translation : {best_translation.final_output}\n\n")
```
